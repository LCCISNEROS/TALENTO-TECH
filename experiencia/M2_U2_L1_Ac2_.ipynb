```python
import skimage
```

```python
# carga imagen de skimage.data
camera = skimage.data.camera()
#explora imagen
print(type(camera))
print(camera.shape)
# mostrar la imagen
skimage.io.imshow(camera)
skimage.io.show()
```

```python
# carga imagen de skimage.data
astronaut = skimage.data.astronaut()
#explora imagen
print(type(astronaut))
print(astronaut.shape)
# mostrar la imagen
skimage.io.imshow(astronaut)
skimage.io.show()
```

```python
import skimage.data
import skimage.io

# carga imagen de skimage.data
image = skimage.data.astronaut()

# mostrar la imagen
skimage.io.imshow(image)
skimage.io.show()
```

```python
import skimage.transform
import skimage.io
import skimage.data

# carga imagen de skimage.data
image = skimage.data.astronaut()

# rotar la imagen 90 grados en sentido antihorario
rotated_image = skimage.transform.rotate(image, angle=30)

# mostrar la imagen
skimage.io.imshow(rotated_image)
skimage.io.show()
```

```python
import skimage.filters
import skimage.io
import skimage.data

# carga imagen de skimage.data
image = skimage.data.astronaut()

# aplicar un filtro gaussiano para suavizar la imagen
smoothed_image = skimage.filters.gaussian(image, sigma=2, channel_axis=-1)

# mostrar la imagen suavizada
skimage.io.imshow(smoothed_image)
skimage.io.show()
```
