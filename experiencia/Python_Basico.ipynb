## 3. Estructuras de datos
Aprenderás las colecciones más usadas:
- **Listas**: ordenadas y mutables
- **Tuplas**: ordenadas e inmutables
- **Conjuntos (set)**: sin orden, sin duplicados
- **Diccionarios**: pares clave–valor

````python
# Listas
numeros = [10, 20, 30]
numeros.append(40)
numeros.append(50)
numeros[1] = 25
print("Lista:", numeros, "Largo:", len(numeros))
````

````python
# Tuplas (como listas que NO cambian)
coordenada = (3, 5)
print(coordenada, coordenada[0])
````

````python
# Conjuntos (sets): eliminan duplicados automáticamente
ciudades = {"Bogotá", "Cali", "Bogotá", "Medellín"}
print(ciudades)
ciudades.add("Barranquilla")
print(ciudades)
````

````python
# Diccionarios (clave -> valor)
persona = {"nombre": "Alex", "edad": 20, "ciudad": "Pasto"}
persona["edad"] = 21
persona["profesion"] = "Diseño"
print(persona)
print("Claves:", list(persona.keys()))
print("Valores:", list(persona.values()))
````

````python
# Comprensiones (listas y diccionarios en una sola línea)
cuadrados = [n*n for n in range(6)]
cuadrados_pares = {n: n*n for n in range(10) if n % 2 == 0}
print(cuadrados)
print(cuadrados_pares)
````

## 4. Control de flujo: condicionales y bucles

````python
# if / elif / else
nota = 85
if nota >= 90:
    print("Excelente")
elif nota >= 70:
    print("Bien")
else:
    print("Necesita repasar")
````

````python
# for y while
for i in range(1, 6):
    print("Número:", i)

n = 5
while n > 0:
    print("Cuenta:", n)
    n -= 1
````

````python
# break y continue
for i in range(10):
    if i == 3:
        continue  # salta el 3
    if i == 8:
        break     # termina el bucle
    print(i, end=" ")
````

## 5. Funciones

````python
def saludo(nombre="mundo"):
    """Devuelve un saludo amable."""
    return f"Hola, {nombre}!"

print(saludo(), saludo("Alex"))
````

````python
# Parámetros variables (*args, **kwargs)
def sumar(*nums):
    return sum(nums)

def mostrar_info(**datos):
    return ", ".join(f"{k}={v}" for k, v in datos.items())

print(sumar(1,2,3,4))
print(mostrar_info(nombre="Alex", edad=21))
````

## 7. Manejo de errores (excepciones)

````python
def dividir(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "No se puede dividir por cero"
    except TypeError:
        return "Tipos no válidos (usa números)"
    finally:
        pass  # lugar para liberar recursos si hiciera falta

print(dividir(10, 2))
print(dividir(10, 0))
print(dividir(10, "x"))
````

## 10. Mini-proyectos (elige 1 o haz ambos)
### A) Juego: Adivina el número
- La computadora elige un número secreto entre 1 y 100.
- Tú intentas adivinarlo; recibe pistas: "mayor/menor".

### B) Presupuesto semanal
- Ingresa ingresos y gastos por categorías.
- Calcula el balance y muestra un resumen.

````python
# A) Adivina el número (versión automática para no depender de input en Colab)
import random

secreto = random.randint(1, 100)
intentos = 0

# Simulamos "intentos" de 1 a 100 hasta acertar, mostrando la lógica
for guess in range(1, 101):
    intentos += 1
    if guess < secreto:
        pass  # pista: es mayor
    elif guess > secreto:
        pass  # pista: es menor
    else:
        print(f"¡Acertado! El número era {secreto}. Intentos: {intentos}")
        break
````

````python
# B) Presupuesto semanal (datos de ejemplo)
ingresos = {"salario": 500, "freelance": 150}
gastos = {"comida": 120, "transporte": 40, "servicios": 90, "ocio": 60}

total_ingresos = sum(ingresos.values())
total_gastos = sum(gastos.values())
balance = total_ingresos - total_gastos

print("Ingresos:", ingresos, "=>", total_ingresos)
print("Gastos:", gastos, "=>", total_gastos)
print("Balance semanal:", balance)
````
