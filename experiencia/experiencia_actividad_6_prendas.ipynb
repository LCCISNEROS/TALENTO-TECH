```python
import keras
# cargue el conjunto de datos Fashion MNIST
fmnist = keras.datasets.fashion_mnist
# cargue la division de entrenamiento y prueba del conjunto de datos Fashion MNIST
(training_images, training_labels), (test_images, test_labels) = fmnist.load_data()
```

```python
import numpy as np
import matplotlib.pyplot as plt
# Puedes poner aquí entre 0 y 59999
index = 1
# imprime la etiqueta y la imagen
np.set_printoptions(linewidth=320)
print(f'Label: {training_labels[index]}')
print(f'image:\n {training_images[index]}')
# Visualiza la imagen
plt.imshow(training_images[index])
```

```python
# Normaliza los valores de pixeles del tren y probar las imágenes
training_images  = training_images / 255.0
test_images = test_images / 255.0
```

```python
# Construir el modelo de clasificacion
model= keras.models.Sequential([keras.layers.Flatten(input_shape=(28,28)), keras.layers.Dense(128, activation="relu"),keras.layers.Dense(10, activation="softmax")])
```

```python
# Compilar el modelo
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.summary()
```

```python
# Entrenar el modelo
history = model.fit(training_images, training_labels, epochs=10)
```

```python
import pandas as pd
```

```python
# graficar el hisotrial de entrenamiento
pd.DataFrame(history.history).plot(grid=True)
```

```python
# Evaluar el modelo en el conjunto de entrenamiento
loss, accuracy = model.evaluate(training_images, training_labels)
print("Pérdida en el conjunto de entrenamiento:", loss)
print("Precisión en el conjunto de entrenamiento:", accuracy)

# Evaluar el modelo con datos no vistos
loss, accuracy = model.evaluate(test_images, test_labels)
print("Pérdida en el conjunto de prueba:", loss)
print("Precisión en el conjunto de prueba:", accuracy)
```

```python
# predict
index = 1
print(f'Label: {test_labels[index]}')
classification = model.predict(test_images[index:index+1])
print(f'Classification:\n {classification.reshape(-1,1)}')
```

# Metricas de clasificaciòn


```python
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

# Obtener las predicciones del modelo en el conjunto de prueba
test_predictions = np.argmax(model.predict(test_images), axis=1)

# Calcular la matriz de confusión
cm = confusion_matrix(test_labels, test_predictions)

# Visualizar la matriz de confusión
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix')
plt.show()
```
